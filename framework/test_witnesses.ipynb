```python
def optimize(W_val, params, bounds):
        """
        Uses Scipy's minimization to find the minimum value of a witness expectation value by
        varying its parameters within given bounds.

        Parameters:
            W_val: a function to get the operator value of a witness
            params: initial guesses for parameters (e.g. theta, alpha, beta)
            bounds: bounds for each of the parameters
        """

        def loss(params):
            """
            Uses the expectation value of the witness as the loss function for minimization
            """
            loss = hard_expec_val(W_val(*params), rho_E).real
            return loss
    
        # Scipy minimization using BFGS gradient descent
        min_W = minimize(loss, params, method="L-BFGS-B", bounds=bounds)

        # return the best-fit params and the minimized expectation value
        return min_W.x, min_W.fun
```
