This Jupyter notebook checks the witness values generated by states_and_witnesses.py against the witness expectation values as if they were hard-coded using the formulas from Paco's thesis. This is done using experimentally-obtained parameter values.

```python
import numpy as np
import states_and_witnesses as sw
```

```python
# Input the filename containing experimental data
filename = "ria_hd_negpi_3_va_trial7/rho_(hd_negpi_3_va-90.0-7).npy"
```

```python
# Pauli Matrices
PAULI_X = np.array([[0,1], [1, 0]])
PAULI_Y = np.array([[0, -1j], [1j, 0]])
PAULI_Z = np.array([[1,0], [0,-1]])
IDENTITY = np.array([[1,0], [0,1]])
```

```python
"""
The following functions get the "hard-coded" operator values of the W3s according to Paco's thesis
instead of doing calculations from first principles as the code does.
"""
def hard_W3_1(theta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.kron(PAULI_Z, PAULI_Z) +
                   np.cos(2*theta)*(np.kron(PAULI_X, PAULI_X) + np.kron(PAULI_Y, PAULI_Y)) +
                   np.sin(2*theta)*(np.kron(PAULI_Z, IDENTITY) + np.kron(IDENTITY, PAULI_Z)))

def hard_W3_2(theta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) - np.kron(PAULI_Z, PAULI_Z) +
                   np.cos(2*theta)*(np.kron(PAULI_X, PAULI_X) - np.kron(PAULI_Y, PAULI_Y)) +
                   np.sin(2*theta)*(np.kron(PAULI_Z, IDENTITY) - np.kron(IDENTITY, PAULI_Z)))

def hard_W3_3(theta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.kron(PAULI_X, PAULI_X) +
                   np.cos(2*theta)*(np.kron(PAULI_Z, PAULI_Z) + np.kron(PAULI_Y, PAULI_Y)) +
                   np.sin(2*theta)*(np.kron(PAULI_X, IDENTITY) + np.kron(IDENTITY, PAULI_X)))

def hard_W3_4(theta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) - np.kron(PAULI_X, PAULI_X) +
                   np.cos(2*theta)*(np.kron(PAULI_Z, PAULI_Z) - np.kron(PAULI_Y, PAULI_Y)) -
                   np.sin(2*theta)*(np.kron(PAULI_X, IDENTITY) - np.kron(IDENTITY, PAULI_X)))

def hard_W3_5(theta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.kron(PAULI_Y, PAULI_Y) +
                   np.cos(2*theta)*(np.kron(PAULI_Z, PAULI_Z) + np.kron(PAULI_X, PAULI_X)) +
                   np.sin(2*theta)*(np.kron(PAULI_Y, IDENTITY) + np.kron(IDENTITY, PAULI_Y)))

def hard_W3_6(theta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) - np.kron(PAULI_Y, PAULI_Y) +
                   np.cos(2*theta)*(np.kron(PAULI_Z, PAULI_Z) - np.kron(PAULI_X, PAULI_X)) -
                   np.sin(2*theta)*(np.kron(PAULI_Y, IDENTITY) - np.kron(IDENTITY, PAULI_Y)))
```

```python
# Load in experimental rho from the file
rho_E = np.load(filename, allow_pickle=True)[0]

def hard_expec_val(hard_val, rho):
    """
    A function to calculate an expectation value using the complex witness operator and complex rho instead
    of Stokes parameters.
    """
    return np.trace(hard_val @ rho)
```

```python
# NOTE: Edit to set the parameter values
theta = 2.35644449

# Initialize a witness object to get the W3 expectation values from the code
W3_obj = sw.W3(rho=rho_E)

# W3_1
print("W3_1 value produced by code:", W3_obj.expec_val(1, theta))
print("W3_1 value produced by formula:", hard_expec_val(hard_W3_1(theta), rho_E))

# W3_2
print("\nW3_2 value produced by code:", W3_obj.expec_val(2, theta))
print("W3_2 value produced by formula:", hard_expec_val(hard_W3_2(theta), rho_E))

# W3_3
print("\nW3_3 value produced by code:", W3_obj.expec_val(3, theta))
print("W3_3 value produced by formula:", hard_expec_val(hard_W3_3(theta), rho_E))

# W3_4
print("\nW3_4 value produced by code:", W3_obj.expec_val(4, theta))
print("W3_4 value produced by formula:", hard_expec_val(hard_W3_4(theta), rho_E))

# W3_5
print("\nW3_5 value produced by code:", W3_obj.expec_val(5, theta))
print("W3_5 value produced by formula:", hard_expec_val(hard_W3_5(theta), rho_E))

# W3_6
print("\nW3_6 value produced by code:", W3_obj.expec_val(6, theta))
print("W3_6 value produced by formula:", hard_expec_val(hard_W3_6(theta), rho_E))
```

```python
# Now compare the operators
# W3_1
print("W3_1 operator produced by code:", W3_obj.W3_1(theta))
print("W3_1 operator produced by formula:", hard_W3_1(theta))

# W3_2
print("\nW3_2 operator produced by code:", W3_obj.W3_2(theta))
print("W3_2 operator produced by formula:", hard_W3_2(theta))

# W3_3
print("\nW3_3 operator produced by code:", W3_obj.W3_3(theta))
print("W3_3 operator produced by formula:", hard_W3_4(theta))

# W3_4
print("\nW3_4 operator produced by code:", W3_obj.W3_4(theta))
print("W3_4 operator produced by formula:", hard_W3_4(theta))

# W3_5
print("\nW3_5 operator produced by code:", W3_obj.W3_5(theta))
print("W3_5 operator produced by formula:", hard_W3_5(theta))

# W3_6
print("\nW3_6 operator produced by code:", W3_obj.W3_6(theta))
print("W3_6 operator produced by formula:", hard_W3_6(theta))
```

```python
"""
"Hard-coded" operator values of the W5s.
"""
def hard_W5_1(theta, alpha):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.kron(PAULI_Z, PAULI_Z) +
                   np.cos(2*theta)*(np.cos(alpha)*(np.kron(PAULI_X, PAULI_X) + np.kron(PAULI_Y, PAULI_Y)) -
                   np.sin(alpha)*(np.kron(PAULI_X, PAULI_Y) - np.kron(PAULI_Y, PAULI_X))) +
                   np.sin(2*theta)*(np.kron(PAULI_Z, IDENTITY) + np.kron(IDENTITY, PAULI_Z)))

def hard_W5_2(theta, alpha):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) - np.kron(PAULI_Z, PAULI_Z) +
                   np.cos(2*theta)*(np.cos(alpha)*(np.kron(PAULI_X, PAULI_X) - np.kron(PAULI_Y, PAULI_Y)) +
                   np.sin(alpha)*(np.kron(PAULI_X, PAULI_Y) + np.kron(PAULI_Y, PAULI_X))) +
                   np.sin(2*theta)*(np.kron(PAULI_Z, IDENTITY) - np.kron(IDENTITY, PAULI_Z)))

def hard_W5_3(theta, alpha, beta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.cos(alpha) * np.cos(beta) * np.kron(PAULI_X, PAULI_X) +
                   np.sin(alpha) * np.sin(beta) * np.kron(PAULI_Y, PAULI_Y) +
                   np.sin(alpha) * np.cos(beta) * np.kron(PAULI_Y, PAULI_X) +
                   np.cos(alpha) * np.sin(beta) * np.kron(PAULI_X, PAULI_Y) +
                   np.cos(2*theta)*(np.kron(PAULI_Z, PAULI_Z) +
                   np.cos(alpha) * np.cos(beta) * np.kron(PAULI_Y, PAULI_Y) +
                   np.sin(alpha) * np.sin(beta) * np.kron(PAULI_X, PAULI_X) -
                   np.sin(alpha) * np.cos(beta) * np.kron(PAULI_X, PAULI_Y) -
                   np.cos(alpha) * np.sin(beta) * np.kron(PAULI_Y, PAULI_X)) + 
                   np.sin(2*theta) * (np.cos(alpha) * np.kron(PAULI_X, IDENTITY) +
                   np.sin(alpha) * np.kron(PAULI_Y, IDENTITY) + np.cos(beta) * np.kron(IDENTITY, PAULI_X) +
                   np.sin(beta) * np.kron(IDENTITY, PAULI_Y)))

def hard_W5_4(theta, alpha):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.kron(PAULI_X, PAULI_X) +
                   np.cos(2*theta)*(np.cos(alpha)*(np.kron(PAULI_Z, PAULI_Z) + np.kron(PAULI_Y, PAULI_Y)) -
                   np.sin(alpha)*(np.kron(PAULI_Y, PAULI_Z) - np.kron(PAULI_Z, PAULI_Y))) +
                   np.sin(2*theta)*(np.kron(PAULI_X, IDENTITY) + np.kron(IDENTITY, PAULI_X)))

def hard_W5_5(theta, alpha):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) - np.kron(PAULI_X, PAULI_X) +
                   np.cos(2*theta)*(np.cos(alpha)*(np.kron(PAULI_Z, PAULI_Z) - np.kron(PAULI_Y, PAULI_Y)) -
                   np.sin(alpha)*(np.kron(PAULI_Y, PAULI_Z) + np.kron(PAULI_Z, PAULI_Y))) -
                   np.sin(2*theta)*(np.kron(PAULI_X, IDENTITY) - np.kron(IDENTITY, PAULI_X)))

def hard_W5_6(theta, alpha, beta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.cos(alpha) * np.cos(beta) * np.kron(PAULI_Z, PAULI_Z) +
                   np.sin(alpha) * np.sin(beta) * np.kron(PAULI_Y, PAULI_Y) -
                   np.sin(alpha) * np.cos(beta) * np.kron(PAULI_Y, PAULI_Z) -
                   np.cos(alpha) * np.sin(beta) * np.kron(PAULI_Z, PAULI_Y) +
                   np.cos(2*theta)*(np.kron(PAULI_X, PAULI_X) +
                   np.cos(alpha) * np.cos(beta) * np.kron(PAULI_Y, PAULI_Y) +
                   np.sin(alpha) * np.sin(beta) * np.kron(PAULI_Z, PAULI_Z) +
                   np.sin(alpha) * np.cos(beta) * np.kron(PAULI_Z, PAULI_Y) +
                   np.cos(alpha) * np.sin(beta) * np.kron(PAULI_Y, PAULI_Z)) + 
                   np.sin(2*theta) * (np.cos(alpha) * np.kron(PAULI_Z, IDENTITY) -
                   np.sin(alpha) * np.kron(PAULI_Y, IDENTITY) + np.cos(beta) * np.kron(IDENTITY, PAULI_Z) -
                   np.sin(beta) * np.kron(IDENTITY, PAULI_Y)))

def hard_W5_7(theta, alpha):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.kron(PAULI_Y, PAULI_Y) +
                   np.cos(2*theta)*(np.cos(alpha)*(np.kron(PAULI_Z, PAULI_Z) + np.kron(PAULI_X, PAULI_X)) +
                   np.sin(alpha)*(np.kron(PAULI_X, PAULI_Z) - np.kron(PAULI_Z, PAULI_X))) +
                   np.sin(2*theta)*(np.kron(PAULI_Y, IDENTITY) + np.kron(IDENTITY, PAULI_Y)))

def hard_W5_8(theta, alpha):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) - np.kron(PAULI_Y, PAULI_Y) +
                   np.cos(2*theta)*(np.cos(alpha)*(np.kron(PAULI_Z, PAULI_Z) - np.kron(PAULI_X, PAULI_X)) +
                   np.sin(alpha)*(np.kron(PAULI_X, PAULI_Z) + np.kron(PAULI_Z, PAULI_X))) -
                   np.sin(2*theta)*(np.kron(PAULI_Y, IDENTITY) - np.kron(IDENTITY, PAULI_Y)))

def hard_W5_9(theta, alpha, beta):
    return 0.25 * (np.kron(IDENTITY, IDENTITY) + np.cos(alpha) * np.cos(beta) * np.kron(PAULI_Z, PAULI_Z) +
                   np.sin(alpha) * np.sin(beta) * np.kron(PAULI_X, PAULI_X) +
                   np.sin(alpha) * np.cos(beta) * np.kron(PAULI_X, PAULI_Z) +
                   np.cos(alpha) * np.sin(beta) * np.kron(PAULI_Z, PAULI_X) +
                   np.cos(2*theta)*(np.kron(PAULI_Y, PAULI_Y) +
                   np.cos(alpha) * np.cos(beta) * np.kron(PAULI_X, PAULI_X) +
                   np.sin(alpha) * np.sin(beta) * np.kron(PAULI_Z, PAULI_Z) -
                   np.sin(alpha) * np.cos(beta) * np.kron(PAULI_Z, PAULI_X) +
                   np.cos(alpha) * np.sin(beta) * np.kron(PAULI_X, PAULI_Z)) + 
                   np.sin(2*theta) * (np.cos(alpha) * np.kron(PAULI_Z, IDENTITY) +
                   np.sin(alpha) * np.kron(PAULI_X, IDENTITY) + np.cos(beta) * np.kron(IDENTITY, PAULI_Z) +
                   np.sin(beta) * np.kron(IDENTITY, PAULI_X)))
```
